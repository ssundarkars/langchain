# Embadding Models

```python
from dotenv import load_dotenv
from sklearn.metrics.pairwise import cosine_similarity
```

```python
load_dotenv()
```

```python
from langchain_openai import OpenAIEmbeddings
embadding = OpenAIEmbeddings(model = "text-embedding-3-large", dimensions=300)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embadding = HuggingFaceEmbeddings(model = "sentence-transformers/all-MiniLM-L6-v2")
```

```python

doc = ["Virat kohili is an Indian bats man with most following world wide.",
       "Ms, Dhoni is most popular Indian wicket skeeper and a bats man, and he is my current favoraite crickater.",
       "Jasprit Bumrah is the most talented fast bowler from Indian team.",
       "Gautam Gambhi is a retired bats man from Indian an and he was my favoraite crickater 10 years ago."]
doc_embed = embadding.embed_documents(doc)
```

```python
query1 = "Who is my favoraite bats man?"
query2 = "Who is most followed world wide?"
query3 = "Name an Indian fast paced bowler."
q1_embed = embadding.embed_query(query1)
q2_embed = embadding.embed_query(query2)
q3_embed = embadding.embed_query(query3)
```

```python
q_similarity = cosine_similarity([q1_embed,q2_embed,q3_embed], doc_embed)

q_similarity
```

```python
doc_attached = [[(seq, value) for seq, value in enumerate(q_sim)] for q_sim in q_similarity ]

doc_attached
```

```python
ans_list =[]
for embed in doc_attached:
    index,value = max(embed, key = lambda p:p[1])
    ans_list.append((index, value ))
```

```python
print(query1, query2,query3, sep= '\n' )
for index, value in ans_list:
    print("Ans: ", doc[index], value)
```
