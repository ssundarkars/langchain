## Promting
### A technique of communication with the LLM model in form of query input, such that the LLM has to take no/minimum assumptions in order to understand and generate the required output.
### Prompting makes response time faster as LLM lookfor the information in limited knowlegde base.


```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
```

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(model = "gemini-3.6-flash" ) #Temp ange 0 - 2  (Strinct - Flexible)
chat_model = init_chat_model(model= "mistral-small-2603", configurable_fields= {
    'temperature' : 1, 
    'max_tokens' : 200 })
```

```python
promptTemplate = PromptTemplate(template = "You are an intelligent space scientist, write a {size} paragraph on topic {topic} explaining it to a {person_type}",
                                input_variables= ['size', 'topic', 'person_type'])
```

```python
size,topic,person_type = 'small', 'blackhole', 'sales man'
```

```python
prompt = promptTemplate.invoke({'size' : size,
                                  'topic' : topic,
                                   'person_type':person_type })
```

```python
response = model.invoke(prompt)
```

```python
response.pretty_print()
```

```python
response_chat = chat_model.invoke(prompt)
```

```python
response_chat.pretty_print()
```

## Structured Output
* There is avery definate need of structured output when working with application which depends on LLM's output to take further actions. 
* Most LLMs support the structured output making easier to process the LLM generated output by other tools in the application.
* Additionally, if any LLM does not support structured output, ther is always a <B>output parser</B> to escape from the situation, but donot guarantees 100% safety :-).

### Structure Formating using...
1. TypedDict (Basic)
2. PyDantic (Pythonic)
3. JSON (Globally used)


```python
from typing import TypedDict, Any, Literal, Annotated, List, Optional
from pydantic import Field, BaseModel , AnyUrl
```

## Typed Dict

```python
class dictAnimal(TypedDict):
    name: str 
    life_span: Annotated[float,"Maximum Life time in Days or Years"]
    life_span_unit: Annotated[Literal['Days', 'Years'], "Unit for 'life_span' key"]
    spend_time: Literal["On Earth", "Under Water", "Under Earth", "Open Sky", "On Water"]
    endangered: Annotated[float ,"Species at risk of extinction. 1 = Very High, 0 = very Low"] 
    travel_type: Literal["Fly", "Walk", "Crawl", "Swim"]     
    fun_fact: Annotated[List[str],"List maximum 2 fun fact about the animal"] 
```

```python
animalPrompt = PromptTemplate(template = "You are an animal expert, provide the required parameters related to {animal} animal",
                              input_variables=["animal"],
                              validate_template = True
                               )
```

```python
prompt3 = animalPrompt.invoke({'animal': 'Whale'})
```

```python
struc_model_chat = chat_model.with_structured_output(dictAnimal)
```

```python
response_chat = struc_model_chat.invoke(prompt3)
```

```python
response_chat
```

## PyDantic
* Default Values
* Type Validation
* Type Coersion  '32' -> 32
* Easy JSON conversion  Pydantic <--> Json
* Regex support 

```python
class pydanticAnimal(BaseModel):
    name: str
    life_span: Annotated[float, "Maximum Life time"]
    life_span_unit: Annotated[Literal['Days', 'Years'], "Unit for 'life_span' key"]
    spend_time: Literal["On Earth", "Under Water", "Under Earth", "Open Sky"]
    endangered: float = Field( ge=0, le=1, description="Species at risk of extinction. 1 = Very High, 0 = very Low" )
    travel_type: Literal["Fly", "Walk", "Crawl", "Swim"] = Field(description="The type of travel")
    fun_fact: Annotated[Optional[List[str]] , "List maximum 2 fun fact about the animal"] 
    source : Annotated[Optional[List[AnyUrl]], "Source of information"]
```

```python
struc_model_chat = chat_model.with_structured_output(pydanticAnimal)
```

```python
response_chat = struc_model_chat.invoke(prompt3)
response_chat
```

## JSON Way (Used Accross Languages/System)

```python
json_student = {
    "title": "Student",
    "description": "More detailed information students",
    "type": "object",
    "properties":{
        "name":{'type':"string"},
        "age":{'type':"integer"},
        "story":{'type': 'string', 'description': "Fictional Story about the character"}
    },
    "required":["name", "age","story"]
}
```

```python
json_model = chat_model.with_structured_output(json_student)
```

```python
response = json_model.invoke("Write about an fictional student character")
response
```
