# Splitter
* Length based
* Text structure based
* Document structure based
* Semantic based

```python
from langchain_classic.text_splitter  import CharacterTextSplitter,RecursiveCharacterTextSplitter
```

```python
charSplitter = CharacterTextSplitter(chunk_size = 300, chunk_overlap = 50 , separator='' )
```

```python
chunks = charSplitter.split_documents(resume_document)
chunks
```

```python
len(chunks)
```

```python
recTextSplitter = RecursiveCharacterTextSplitter(
            chunk_size = 400,
            chunk_overlap = 50,
            separators= ['\n\n', '\n']
)
```

```python
chunks = recTextSplitter.split_documents(resume_document)
```

```python
chunks
```

## Document Text splitter

```python
from langchain_classic.text_splitter import Language

docSplitter = RecursiveCharacterTextSplitter.from_language(
    language= Language.PYTHON,
    chunk_size = 100,
    chunk_overlap = 0
)
```

```python
code_chunks = docSplitter.split_documents(pycode_text)
code_chunks
```
