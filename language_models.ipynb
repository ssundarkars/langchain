## Hugging Face (Open Source)

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint


llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation",
    temperature= 1.5
)

model = ChatHuggingFace(llm=llm)

result = model.invoke("Hello my name is Beast." )

print(result.content)
```

### Langchain provides support for most of the Model providers, all the model providers has their own class implemented class "\<provider\>" from "langchain_\<provider\>"
### Additionally langchain has given support for common module that can works will all model provider i.e. <B>init_chat_models</B> from langchain.<B>chat_models </B>.

```python
from langchain.chat_models import init_chat_model
chat_model = init_chat_model(model= "mistral-small-2603")
```

```python
response = chat_model.invoke("Hello, My name is Beast.")

response.content
```

```python
response = chat_model.invoke("Tell something funny about my name." )

response.pretty_print()
```
